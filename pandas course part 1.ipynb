### ADDING & MODIFYING COLUMNS

```python

df = pd.read_csv("./data_f/students.csv")
```

```python
# Add a new column
df["is_Scholarship_Eligible"] = df["Marks"] >= 90
df
```

```python
# Modifying an existing column
df["Marks"] = df["Marks"] + 2
df
```

```python
# apply() - custom logic per row
def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    else:
        return "C"

df["Grade"] = df["Marks"].apply(get_grade)
df
```

### GROUPBY & AGGREGATION

```python
df = pd.read_csv("./data_f/students.csv")
df
```

```python
# Basic Groupby
mean_values = df.groupby("Course")["Marks"].mean()
mean_values
```

```python
max_values = df.groupby("Course")["Marks"].max()
max_values
```

```python
min_values = df.groupby("Course")["Marks"].min()
min_values
```

```python
# multiple aggregation with agg()

results = df.groupby("Course")["Marks"].agg(["mean","max","min", "count"])
results
```

```python
results2 = df.groupby("Course")["Attendance"].agg(["mean","max","min", "count"])
results2
```

### MERGING & JOINING DATAFRAMES

```python

students = pd.DataFrame({
    "ID": [1,2,3],
    "Name": ["Nor", "Fatin", "Nabila"]
})
display(students)

scores = pd.DataFrame({
    "ID": [1,2,4],
    "Scores": [85,92,78]
})
display(scores)
```

```python
# Inner join - keep only matches

results = pd.merge(students, scores, on="ID", how="inner")
results
```

```python
# Left join - keep all left rows

results2 = pd.merge(students, scores, on="ID", how="left")
results2
```

```python
# Right join - keep all the right rows
results3 = pd.merge(students, scores, on="ID", how="right")
results3
```

```python
# Outer join - keep every row from both Table
results4 = pd.merge(students, scores, on="ID", how="outer")
results4
```
